```python
import numpy as np
from matplotlib import pyplot as plt
import os, glob
```

```python
files = glob.glob('*.npz')
reference = np.load('TIM_loading_per_channel_FiducialAPRA.npz')
print(files)
```

```python
def color_selector(f):
    if "Fiducial" in f:
        color = 'black'
    elif "NoBlueLeaks" in f:
        color = 'Orange'
    elif "ReflectiveNIR" in f:
        color = 'Blue'
    elif "AbsorptiveNIR" in f:
        color = 'Green'
    elif "RealisticNIR" in f:
            color = 'Red'
    else:
        color = None
    return color
```

```python
plt.figure()
for i, f in enumerate(files):
    color = color_selector(f)
    ls = '--' if "Proposed" in f else ':'
    data = np.load(f)
    label = f.split('_')[-1].split('.')[0]
    plt.plot(data['umLW'], data['fW_LW'] / reference['fW_LW'], label=label, color=color, ls=ls)
    plt.plot(data['umSW'], data['fW_SW'] / reference['fW_SW'], color=color, ls=ls)

# plt.xscale('log')
plt.xlabel('Wavelength (um)')
plt.legend(ncol=2, loc='lower center', fontsize=8)
plt.ylabel('Detector Loading / Fiducial 2023APRA')
plt.show()
```
